## Calcul des statistiques descriptives

### Mode(manuellement)

```python
print(" Mode ")
for column, values in dictionnary_data.items():
    # compter combien de fois chaque valeur apparait
    frequences = {}
    for value in values:
        if value in frequences:
            frequences[value] += 1
        else:
            frequences[value] = 1

    # trouver la frequence maximale
    max_freq = 0
    for freq in frequences.values():
        if freq > max_freq:
            max_freq = freq

    # recuperer toutes les valeurs qui ont cette frequence 
    modes = []
    for value, freq in frequences.items():
        if freq == max_freq:
            modes.append(value)
    modes.sort()

    print(f"{column:<25} {modes}")
```

## Normalisation

### Methode MIN MAX(manuellement)

```python
print(" Normalisation MIN MAX ")

# Creer un dictionnaire comme dictionnary_data mais sans trier les valeurs pour pouvoir le comparer avec le tableau pandas
with open('datasets/breast.csv', 'r') as f:
    file = csv.reader(f, delimiter=',')
    columns = next(file)

    dictionnary_data_non_trie = {column: [] for column in columns} 

    for line in file:
      for i, value in enumerate(line):
        dictionnary_data_non_trie[columns[i]].append(float(value))
normalized_data = {}

for column, values in dictionnary_data_non_trie.items():
    min_value = min(values)
    max_value = max(values)
    if min_value != max_value: # Condition pour éviter la division par zéro
      normalized_data[column] = [(value - min_value) / (max_value - min_value) for value in values]
    else:
      print(f"{column}: normalisation impossible (toutes les valeurs sont identiques)")

tableau = pd.DataFrame(normalized_data) # Afficher les données normalisées sous forme de tableau pandas
tableau.head(len(data))
```

### Methode MIN MAX(utilisation de pandas)

```python
print(" Normalisation MIN MAX ")

data_numeriques = data.select_dtypes(include='number')   # extraire les colonnes numeriques

data_norm = (data_numeriques - data_numeriques.min()) / (data_numeriques.max() - data_numeriques.min())
data_norm

```

### Methode Z-SCORE(manuellement)

```python
print(" Normalisation Z-SCORE ")
for column, values in dictionnary_data_non_trie.items():
    # Recalcul de la moyenne
    moyenne = sum(values) / len(values)
    # Calcul de l'écart type
    variance = sum((value - moyenne) ** 2 for value in values) / (len(values)-1)
    ecart_type = math.sqrt(variance)

    if ecart_type != 0:
        normalized_data[column] = [(value - moyenne) / ecart_type for value in values]
    else:
        print(f"{column}: normalisation impossible (l'écart type est nul)")

tableau = pd.DataFrame(normalized_data) # Afficher les données normalisées sous forme de tableau pandas
tableau.head(len(data))
```

### Methode Z-SCORE(utilisation de pandas)

```python
print(" Normalisation Z-SCORE ")

data_numeriques = data.select_dtypes(include='number')   # extraire les colonnes numeriques

data_norm = (data_numeriques - data_numeriques.mean()) / (data_numeriques.std())
data_norm

```

### Valeurs manquantes

```python
#  decoder les bytes en texte
for c in data.select_dtypes(include=object):
    data[c] = data[c].apply(lambda x: x.decode() if isinstance(x, bytes) else x)

# transformer '?' en NaN
data = data.replace('?', np.nan)

missingValues = data.isnull().sum()
print("Valeurs manquantes")
missingValues
```

### Remplacemenet des valeurs manquantes par la moyenne ou le mode

```python
# valeurs numeriques remplacees par la moyenne
num_cols = data.select_dtypes(include='number').columns
data[num_cols] = data[num_cols].fillna(data[num_cols].mean())

# valeurs categorielles remplacees par le mode
cat_cols = data.select_dtypes(exclude='number').columns
for c in cat_cols:
    data[c] = data[c].fillna(data[c].mode()[0])
```

### Affichage des valeurs manquantes apres le remplacement

```python
missingValues = data.isnull().sum()
print("Valeurs manquantes apres remplacement")
missingValues
```

### Supression de la colonne TBG car toutes ses lignes sont manquantes

## Normalisation

### Methode MIN MAX(utilisation de pandas)

```python
print(" Normalisation MIN MAX ")

data_numeriques = data.select_dtypes(include='number')   # extraire les colonnes numeriques

data_norm = (data_numeriques - data_numeriques.min()) / (data_numeriques.max() - data_numeriques.min())
data_norm

```

### Methode Z-SCORE(utilisation de pandas)

```python
print(" Normalisation Z-SCORE ")

data_numeriques = data.select_dtypes(include='number')   # extraire les colonnes numeriques

data_norm = (data_numeriques - data_numeriques.mean()) / (data_numeriques.std())
data_norm

```
