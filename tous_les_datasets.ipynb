### afficher la mediane(manuellement)

```python

print(" Mediane ")
#verification si le nombre de ligne est pair ou impair
if len(data) % 2 == 0:
    median_index1 = math.ceil(len(data) / 2) - 1 #pour avoir lentier superieur
    median_index2 = math.ceil(len(data) / 2) 
    for column, values in dictionnary_data.items():
        median = (values[median_index1] + values[median_index2]) / 2
        print(f"{column:<25} {median:>10}")
else:
    median_index = math.ceil(len(data) / 2) - 1
    for column, values in dictionnary_data.items():
        median = values[median_index]
        print(f"{column:<25} {median:>10}")
```

```python
median_values = data.median(numeric_only=True)
print("Mediane")
print(median_values)
```

### afficher Q1(manuellement)

```python
print(" Q1 ")
#verification si le nombre de ligne est pair ou impair
if len(data) % 2 == 0:
    Q1_index1 = math.ceil(len(data) / 4) - 1 #pour avoir lentier superieur
    Q1_index2 = math.ceil(len(data) / 4 ) 
    for column, values in dictionnary_data.items():
        median = (values[Q1_index1] + values[Q1_index2]) / 2
        print(f"{column:<25} {median:>10}")
else:
    Q1_index = math.ceil(len(data) / 4)
    for column, values in dictionnary_data.items():
        median = values[Q1_index]
        print(f"{column:<25} {median:>10}")
```
